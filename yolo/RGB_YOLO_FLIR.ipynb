```python
"""
Phase 1, Config A: RGB-only YOLOv8n baseline on FLIR-aligned.
FULL RUN.
"""

from ultralytics import YOLO

DATA_CONFIG = r"C:\Users\LENOVO\Documents\Thermal_Fusion_Paper\configs\flir_aligned.yaml"
EPOCHS = 100
IMG_SIZE = 640
BATCH = 16
PROJECT = r"C:\Users\LENOVO\Documents\Thermal_Fusion_Paper\results"
RUN_NAME = "config_a_rgb_baseline"

model = YOLO("yolov8n.pt")

results = model.train(
    data=DATA_CONFIG,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH,
    project=PROJECT,
    name=RUN_NAME,
    patience=20,   # stop early if val mAP plateaus for 20 epochs
    val=True,
    plots=True,
)

print("Training complete. Best weights at:")
print(f"  {PROJECT}\\{RUN_NAME}\\weights\\best.pt")
```
